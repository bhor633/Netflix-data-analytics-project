```python
# Import Libraries

import pandas as pd
import numpy as np
```

```python
# String (str) columns to object 
df = df.astype({
    'Customer_ID': 'object',
    'Customer_Name': 'object',
    'Title': 'object',
    'Content_Type': 'object',
    'Genre': 'object',
    'Country': 'object',
    'Subscription_Plan': 'object',
    'Device': 'object',
    'Payment_Method': 'object',
    'Account_Status': 'object'
})
```

```python
df.info()
```

```python
# Statistical Summary
df.describe()
```

```python
# Missing Values
df.isnull().sum()
```

```python
# Duplicate Records

df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
# Dirty values - N/A,NULL,BLANK,SPACES

df.replace(
    ['N/A','NULL','',' '],np.nan,inplace=True
)
```

```python
df.isnull().sum()
```

```python
# Extra spaces remove

df=df.apply(lambda x: x.str.strip() if x.dtype =="object" else x)
```

```python
# standardize Text

df["Country"] = df["Country"].str.title()

df["Genre"] = df["Genre"].str.title()

df["Subscription_Plan"] = (df["Subscription_Plan"].str.title())

df["Account_Status"] = ( df["Account_Status"].str.title())
```

```python
df['Country'].value_counts()
```

```python
# Numaric columns 

numeric_cols = ['Age', 'Watch_Hours',
                'Monthly_Fee', 'Movies_Watched',
                'Episodes_Watched', 'Avg_Rating',
                'Monthly_Sessions', 'Download_Count']
```

```python
for col in numeric_cols:
     df[col]=pd.to_numeric(df[col],errors='coerce')
```

```python
df.describe()
```

```python
df = df[(df['Age']>=18)&(df['Age']<=100)]
```

```python
df['Age'].describe()
```

```python
# Nagetive values

df = df[df['Watch_Hours']>=0]
df = df[df['Monthly_Fee']>=0]
```

```python
# data date cleaning

df['Join_Date'] = pd.to_datetime(df['Join_Date'],errors='coerce')
df['Last_Watch_Date'] = pd.to_datetime(df['Last_Watch_Date'],errors='coerce')
```

```python
df[["Join_Date", "Last_Watch_Date"]].isnull().sum()
```

```python
#Missing values Handle

df["Age"] = df["Age"].fillna(df["Age"].mean())

df["Watch_Hours"] = df["Watch_Hours"].fillna(
    df["Watch_Hours"].mean()
)

df["Monthly_Fee"] = df["Monthly_Fee"].fillna(
    df["Monthly_Fee"].mean()
)
```

```python
df["Payment_Method"] = df["Payment_Method"].fillna("Unknown")

df["Device"] = df["Device"].fillna("Unknown")

df["Genre"] = df["Genre"].fillna("Unknown")
```

```python
# Feature Engineering

# Customer Revenue
df["Customer_Revenue"] = (
    df["Monthly_Fee"] * df["Monthly_Sessions"]
)
```

```python
df.head()
```

```python
# Engagement score

df["Engagement_Score"] = (
    df["Watch_Hours"] +
    df["Monthly_Sessions"] +
    df["Download_Count"]
)
```

```python
df.head()
```

```python
# Watch Category

# Watch Category

bins = [0, 20, 50, 100, np.inf]
labels = ['Low', 'Medium', 'High', 'Very High']

df['Watch_Category'] = pd.cut(
    df['Watch_Hours'],
    bins=bins,
    labels=labels
)
```

```python
df.head()
```

```python
 df["Watch_Category"].value_counts()
```

```python
 # Customer Age Group

bins = [17, 25, 35, 50, 100]
labels = ['18-25', '26-35', '36-50', '51+']

df['Age_Group'] = pd.cut(
    df['Age'],
    bins=bins,
    labels=labels,
    include_lowest=True
)
```

```python
df['Age_Group'].value_counts()
```

```python
#  Account Active flag
df["Active_Flag"] = np.where(df["Account_Status"] == "Active", 1, 0)
```

### Remove rows with remaining null values
37 customers have `Watch_Hours = 0` (so `Watch_Category` is null) and 1 customer has a missing `Country`. These 38 rows are removed so that Python, SQL and Power BI all use the same dataset (957 rows).

```python
# Remove rows that still contain null values
before = len(df)
df = df.dropna()
print(f"Rows before: {before} | removed: {before - len(df)} | after: {len(df)}")
```
